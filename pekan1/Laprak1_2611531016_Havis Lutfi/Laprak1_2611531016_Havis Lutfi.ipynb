## 2.1 Dasar Teori
Pada praktikum kita melakukan pengenalan dengan membahas penggunaan bahasa pemograman python menggunakan kode editor Visual Studio Code, yang dimana kode yang telah dibuat akan diunggah (push) kedalam repository github untuk disimpan dan dikelola dengan menggunakan fitur aplikasi VS Code atau menggunakan perintah dasar git pada terminal.

1. Python.

   Python adalah bahasa pemrograman tingkat tinggi yang terkenal dengan sintaksnya yang ringkas serta mudah dipahami.
2. Visual Studio Code (VS Code).

   Visual Studio Code merupakan sebuah Code Editor yang dikembangkan oleh Microsoft. VS Code mendukung berbagai bahasa pemrograman termasuk Python serta dilengkapi dengan fitur Integrated Terminal, Syntax Highlighting, IntelliSense (pembaca sintaks otomatis), dan ekstensi bawaan untuk integrasi dengan Git.
3. Git.

   Git merupakan Sistem kontrol versi (Version Control System / VCS), melacak sejarah perubahan saat orang dan tim berkolaborasi dalam proyek bersama. Saat pengembang membuat perubahan pada proyek, versi proyek sebelumnya dapat dipulihkan kapan saja.
4. Github.

   GitHub adalah platform yang menyediakan layanan hosting untuk repositori Git. GitHub memungkinkan pengembang untuk menyimpan kode program secara online di internet , mempermudah kolaborasi antar tim, serta menjadi media dokumentasi dan portofolio proyek perangkat lunak.
   a. Repository.

      Repositori adalah elemen paling dasar dari GitHub. Ini adalah tempat di mana Anda dapat menyimpan kode Anda, file Anda, dan riwayat revisi setiap file. Repositori dapat memiliki banyak kolaborator dan dapat bersifat publik atau swasta.
5. Commit and Push.

   a. Commit.

      Proses penandaan perubahan dari file proyek untuk disimpan secara permanen ke dalam riwayat repositori lokal disertai pesan penjelasan.
      
   b. Push.
   
      Perintah untuk mengunggah dan memindahkan seluruh rekaman commit dari repositori lokal ke repositori remote yang ada di server GitHub.

### 2.2.2 Mengunggah kode ke Github"

Untuk mengunggah kode yang telah dibuat di VS Code, ikuti tahapan berikut:

1. Buatlah repository github
2. Buka terminal pada vs code di kode yg ingin di unggah
3. Eksekusi Perintah Git pada Terminal

contoh perintah:

```python
git config --global user.name "usernamegithubsaya"
git config --global user.email "emailgithubsaya@gmail.com"
git init
git add .
git commit -m "pesan commit"
git branch -M main
git remote add origin https://github.com/username/nama-repo.git
git push -u origin main
```

Penjelasan perintah di atas:

1. git config --global user.name "usernamegithubsaya" digunakan untuk konfigurasi username github
2. git config --global user.email "emailgithubsaya@gmail.com" digunakan untuk konfigurasi email github
3. git init digunakan untuk membuat repository Git pada folder proyek.
4. git add . digunakan untuk menambahkan seluruh file code ke staging area.
5. git commit digunakan untuk menyimpan perubahan dengan sebuah pesan.
6. git branch -M main digunakan untuk menentukan branch utama dengan nama main.
7. git remote add origin digunakan untuk menghubungkan repository lokal dengan repository GitHub.
8. git push digunakan untuk mengunggah kode dari komputer ke GitHub.
