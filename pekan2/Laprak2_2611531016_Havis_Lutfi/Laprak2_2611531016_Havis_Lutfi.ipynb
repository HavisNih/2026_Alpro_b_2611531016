## 2.1 Dasar Teori
Pada praktikum ini mahasiswa melakukan pengenalan dan mengimplementasikan tentang variabel, tipe data (seperti string, boolean, integer, float, dan complex, ), konversi tipe data, konstanta, hingga sistem input dari pengguna.

1. Tipe Data Numeric.

   Pada python terdapat beberapa tipe data numeric yaitu integer untuk bilangan bulat (100), tipe data float untuk bilangan pecahan atau desimal (10.2345), dan ada juga tipe data complex untuk bilangan imajiner atau kompex (100+3j).
2. Tipe data string.

   Tipe data string (str) adalah tipe data untuk menyimpan teks atau karakter. Pada python dapat digunakan dengan menggunakan tanda peting tunggal ('), petik ganda ("), petik tiga (""") untuk teks baris lebih dari 1.
3. Tipe Data Boolean.

   Tipe data boolean adalah tipe data yang hanya mengeluarkan True dan False. Tipe ini biasanya digunakan untuk melakukan perbandingan terhadap sesuatu sehingga dibutuhkan hasil apakah dia true atau false.
4. Fungsi input dan konversi data.

   a. input().

      FUngsi input() digunakan untuk memungkinkan program untuk mendapatkan masukkan dari pengguna. fungsi input ini akan menghasilkan nilai yang bertipe data string (str).

   b. Konversi tipe data.
      
      Fungsi ini biasanya digunakan untuk mengubah tipe data yang diberikan oleh pengguna "input()" yang memberikan nilai bertipe string. Untuk mengubah tipe data kita dapat menggunakan " int() " untuk mengubah tipe data menjadi integer (bilangan bulat), " float() " untuk menjadikannya bilangan desimal, atau bisa juga menggunakan " str() " jika kita ingin mengubah suatu tipr data menjadi string.
5. Variabel.

   Variabel adalah sebuah wadah tempat di mana kita bisa memasukkan sesuatu di dalamnya, yaitu memasukkan atau menyimpan data. Untuk menyimpan nilai ke dalam variabel kita perlu menggunakan tanda "=" . contoh " panjang_1016 = 27  ".
6. Konstanta.

   Konstanta yaitu variabel yang nilainya tidak boleh berubah. contohnya pada " PI_1016: Final = 3.14  " ini menandai bahwa nilai dari PI tidak boleh berubah yg bernilai 3.14.

### 2.2.1 Program Numerik
Pada praktikum ini, kode python dengan file bernama numeric_2611531016 digunakan untuk mengetahui tipe data numerik dari 3 tipe data yaitu integer, float, dan complex.

```python
# Buat file dengan nama Numerik_2611531016.py
# create a variable with integer value.
a_1016 = 100
print ( "The type of variable having value", a_1016, "is", type (a_1016))

#create a variable with float value.
b_1016 = 10.2345
print ("The type of variable having value", b_1016, "is", type (b_1016))

#create a variable with complex value.
c_1016 = 100+3j
print ("The type of variable having value", c_1016, "is", type (c_1016))
```

Program ini menghasilkan:

The type of variable having value 100 is <class 'int'>

The type of variable having value 10.2345 is <class 'float'>

The type of variable having value (100+3j) is <class 'complex'>


**Penjelasan :**
1. Program mendefinisikan 3 variabel yang memiliki tipe data numerik yang berbeda yaitu a_1016, b_1016, dan c_1016.
2. fungsi print() berisikan program untuk mencetak tipe data dari masing variabel.
3. fungsi type() digunakan untuk mengetahui tipe data dari masing masing variabel.

### 2.2.2 Program Boolean

Pada praktikum ini, kode python dengan file bernama numeric_2611531016 digunakan untuk menentukan apakah mahasiswa lulus dengan peringkat cumlaude atau bukan dengan membandingkan batas nilai kelulusan dan batas nilai serta menyatakan status kelulusan dengan menggunakan boolean.

code:

```python
# Buat file dengan nama Boolean_2611531016.py
# Nama variabel ditambah 4 digit nim terakhir
# Deklarasi variabel dengan tipe data boolean
is_lulus_1016 = True
is_cumlaude_1016 = True

# Menggunakan boolean
nilai_1016 = 85
batas_lulus_1016 = 75

#Menentukan nilai boolean dari kondisi
status_kelulusan_1016 = nilai_1016 >= batas_lulus_1016 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai: ", nilai_1016)
print("Apakah lulus?: ", status_kelulusan_1016)
if is_lulus_1016 and is_cumlaude_1016:
    print("Selamat, Anda lulus dengan peringkat cumlaude!")
```

Program diatas menghasilkan:

=== Check Kelulusan ===

Nilai:  85             

Apakah lulus?:  True

Selamat, Anda lulus dengan peringkat cumlaude!         

Penjelasan :

1. Variabel is_lulus_1016 dan is_cumlaude dibuat untuk mentakan bahwa ia lulus (True) dan cumlaude (True).
2. variabel nilai_1016 dan batas_lulus_1016 dibuat untuk menyatakan berapa nilai dan berapa batas kelulusan.
3. Variabel status_kelulusan_1016 dibuat untuk menyimpan hasil berupa boolean berdasarkan perbandingan antara variabel nilai_1016 dan batas_lulus_1016 yang akan menghasilkan true atau false berdasarkan nilai dari variabel nilai_1016.
4. Fungsi print ("=== Check Kelulusan ===") untuk mencetak "=== Check Kelulusan ===" sebagai judul.
5. Fungsi print("Nilai: ", nilai_1016) untuk untuk menampilkan berapa nilai mahasiswa.
6. Fungsi print("Apakah lulus?: ", status_kelulusan_1016) adalah menentukan status kelulusan berdasarkar nilai variabel status_kelulusan_1016.
7. Fungsi if is_lulus_1016 and is_cumlaude_1016: untuk untuk melakukan kepututsan dengan membandingkan 2 variabel, lalu menggunakan operator logika "and" yang membuat kedua variabel harus bernilai true.
8. dan terakhir fungsi print("Selamat, Anda lulus dengan peringkat cumlaude!") akan dijalankan jika keputusan pada pada "if" bernilai benar yg menghasilkan teks bahwa anda lulus dengan peringkat cumlaude.
