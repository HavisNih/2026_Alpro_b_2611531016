## 2.1 Dasar Teori
Pada praktikum ini mahasiswa melakukan pengenalan dan mengimplementasikan tentang variabel, tipe data (seperti string, boolean, integer, float, dan complex, ), konversi tipe data, konstanta, hingga sistem input dari pengguna.

1. Tipe Data Numeric.

   Pada python terdapat beberapa tipe data numeric yaitu integer untuk bilangan bulat (100), tipe data float untuk bilangan pecahan atau desimal (10.2345), dan ada juga tipe data complex untuk bilangan imajiner atau kompex (100+3j).
2. Tipe data string.

   Tipe data string (str) adalah tipe data untuk menyimpan teks atau karakter. Pada python dapat digunakan dengan menggunakan tanda peting tunggal ('), petik ganda ("), petik tiga (""") untuk teks baris lebih dari 1.
3. Tipe Data Boolean.

   Tipe data boolean adalah tipe data yang hanya mengeluarkan True dan False. Tipe ini biasanya digunakan untuk melakukan perbandingan terhadap sesuatu sehingga dibutuhkan hasil apakah dia true atau false.
4. Fungsi input dan konversi data.

   a. input().

      FUngsi input() digunakan untuk memungkinkan program untuk mendapatkan masukkan dari pengguna. fungsi input ini akan menghasilkan nilai yang bertipe data string (str).

   b. Konversi tipe data.
      
      Fungsi ini biasanya digunakan untuk mengubah tipe data yang diberikan oleh pengguna "input()" yang memberikan nilai bertipe string. Untuk mengubah tipe data kita dapat menggunakan " int() " untuk mengubah tipe data menjadi integer (bilangan bulat), " float() " untuk menjadikannya bilangan desimal, atau bisa juga menggunakan " str() " jika kita ingin mengubah suatu tipr data menjadi string.
5. Variabel.

   Variabel adalah sebuah wadah tempat di mana kita bisa memasukkan sesuatu di dalamnya, yaitu memasukkan atau menyimpan data. Untuk menyimpan nilai ke dalam variabel kita perlu menggunakan tanda "=" . contoh " panjang_1016 = 27  ".
6. Konstanta.

   Konstanta yaitu variabel yang nilainya tidak boleh berubah. contohnya pada " PI_1016: Final = 3.14  " ini menandai bahwa nilai dari PI tidak boleh berubah yg bernilai 3.14.

### 2.2.5 Program Konstanta_2611531016

Program menghitung luas lingkaran dengan menetapkan PI sebagai konstanta yang tidak bisa diubah.

```python
# Buat file dengan nama Konstanta_2611531016.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# Nama variabel ditambah 4 digit terakhir nim

from typing import Final
PI_1016: Final = 3.14
print("pi: %f" % (PI_1016))
jari_1016 = float(input("Masukkan nilai jari-jari: "))
luas_1016 = PI_1016 * jari_1016 * jari_1016
print("Luas lingkaran dengan jari jari %.2f adalah %.2f" % (jari_1016, luas_1016))
```

Hasil :

  

Penjelasan :

1. Mengimpor Final dari modul typing untuk menetapkan variabel PI_1016 bernilai 3.14 sebagai konstanta.
2. Tanda %f berarti angka yang ditampilkan berubah menjadi desimal dengan 6 angka dibelakang koma.
3. Variabel jari_1016 menerima input jari jari dari pengguna dan dikonversi menjadi tipe data float.
4. Menghitung luas lingkaran dengan mengkalikan PI_1016 dengan jari_1016 dan jari_1016.
5. mencetak nilai luas lingkaran dengan fungsi print().
6. Tanda %.2f berarti angka yang ditampilkan dibulatkan menjadi dua angka di belakang koma.

### 2.2.5 Program tugas2_2611531016

Program menerima input dari pengguna berupa data diri dan nilai, lalu menampilkan tipe data dari masing-masing masukan inputan pengguna. Diakhir program menamilkan apakah berdasarkan nilai mahasiswa tersebut ia lulus.

```python
from typing import Final, final

sinyal_1016 = 100+3j
batasnilai_1016 : Final = 75.0

print("=== SISTEM REGISTRASI PRAKTIKAN ALPRO 2026 ===")
nama_1016 = input("Masukkan Nama Mahasiswa : ")
kelamin_1016 = input("Masukkan Jenis Kelamin (L/P) : ")
umur_1016 = int(input("Masukkan Umur Mahasiswa : "))
skor_1016 = float(input("Masukkan Skor Tes Awal : "))
alamat_1016 = """
Jl. Kampus Unand,
Kecamatan Pauh,
Kota Padang"""

print("\n=== DATA PRAKTIKAN & HASIL PEMERIKSAAN ===")
print("Nama Mahasiswa : ", nama_1016, "| TIPE: ", type(nama_1016) )
print("Jenis Kelamin : ", kelamin_1016, "| TIPE: ", type(kelamin_1016) )
print("Alamat Domisili:", alamat_1016, "| TIPE: ", type(alamat_1016) )
print("Umur : ", umur_1016, "tahun | TIPE: ", type(umur_1016) )
print("Skor Tes Awal : ", skor_1016, "| TIPE: ", type(skor_1016) )
print("ID Token Sinyal :", sinyal_1016, "| TIPE: ", type(sinyal_1016) )

Lulus_1016 = skor_1016 >= batasnilai_1016

print("\n=== STATUS KELULUSAN PRAKTIKUM ===")
print("Batas Minimum Nilai :", batasnilai_1016)
print("Apakah Dinyatakan Lulus? :", Lulus_1016, "| TIPE: ", type(Lulus_1016) )
```

Hasil :

 

Penjelasan :
