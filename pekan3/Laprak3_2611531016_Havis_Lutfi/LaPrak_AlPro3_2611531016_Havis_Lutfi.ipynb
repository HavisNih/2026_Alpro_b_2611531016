## 2.1 Dasar Teori
Pada praktikum ini mahasiswa melakukan pengenalan dan mengimplementasikan tentang operator aritmatika, perbandingan (comparison), logika, penugasan (asignment), keanggotaan (membership), identitas (identity), dan bitwise.

1. Operator

    Operator Python adalah simbol atau kata kunci khusus yang melakukan operasi pada variabel dan nilai, berfungsi sebagai blok bangunan mendasar untuk manipulasi data, perbandingan, dan kontrol logis dalam bahasa..
2. Operator Aritmatika.

    Operator aritmatika digunakan dengan nilai numerik untuk melakukan operasi matematika umum seperti:  
    - a. Penjumlahan (+).  
    - b. pengurangan (-).  
    - c. Perkalian (*).  
    - d. Pembagian (/).  
    - e. Pembagian bulat (//).  
    - f. Modulus/sisa bagi (%).  
    - g. pangkat (**).  
3. Operator Perbandingan (Comparison).

    Operator perbandingan digunakan untuk membandingkan dua nilai dan mengeluarkan `True` atau `False` berdasarkan perbandingan, beberapa operator perbandingan:  
    - a. Sama dengan (==).  
      Bernilai `TRUE` apabila operand kiri sama dengan operand kanan, bernilai `FALSE` jika operand kiri tidak sama dengan operand kanan.  
    - b. Tidak sama dengan (!=).  
      Bernilai `TRUE` apabila operand kiri tidak sama dengan operand kanan, bernilai `FALSE` jika operand kiri sama dengan operand kanan.  
    - c. Lebih besar (>).  
      Bernilai `TRUE` apabila operand kiri lebih besar dari operand kanan, bernilai `FALSE` jika operand kiri lebih kecil dari operand kanan.  
    - d. Lebih kecil (<).  
      Bernilai `TRUE` apabila operand kiri lebih kecil dari operand kanan, bernilai `FALSE` jika operand kanan lebih kecil dari operand kanan.  
    - e. Lebih besar atau sama dengan (>=).  
      Bernilai `TRUE` apabila operand kiri lebih besar dari atau sama dengan operand kanan, bernilai `FALSE` jika operand kiri lebih kecil dari atau sama dengan operand kanan.  
    - f. Lebih kecil atau sama dengan (<=).  
      Bernilai `TRUE` apabila operand kiri lebih kecil dari atau sama dengan operand kanan, bernilai `FALSE` jika operand kiri lebih besar dari atau sama dengan operand kanan.  
    - g. Perbandingan berantai.  
      Perbandingan berantai ini emungkinkan penggunaan beberapa operator perbandingan secara berurutan dalam satu ekspresi. Akan mengasilkan `TRUE` atau `FALSE` tergantung operasi yang dilakukan. 
4. Operator Logika.  
    Operator logika pada python adalah operator logika untuk melakukan operasi data boolean. Operator tersebut mengoperasikan data boolean, yaitu data `TRUE` dan `FALSE`, data `TRUE` bernilai benar dan data `FALSE` bernilai salah. Terdapat 4 jenis operator logika yaitu:  
    - a. Konjungsi AND.  
          Dapat dilakukan menggunakan function `and`. Nilai kebenaran konjungsi AND adalah benar `TRUE` ketika kedua operand (nilai) bernilai benar.  
    - b. Disjungsi OR.  
          Dapat dilakukan menggunakan fungsi `or`. Disjungsi OR mempunyai nilai kebenaran salah saat kedua operand bernilai salah.  
    - c. Negasi NOT.  
          Dapat dilakukan menggunakan fungsi `not`. Negasi not digunakan untuk membalik nilai Boolean dari suatu kondisi.   
    - d. Disjungsi eksklusif XOR.  
          Adalah operasi logika dari OR Eksklusif. Kita dapat menggunakan fungsi `!=` untuk menggunakan Disjungsi eksklusif XOR. Operasi XOR menghasilkan `TRUE` jika kedua nilai berbeda.  
    
5. Operator Penugasan (Asignment).

   Operator penugasan digunakan untuk menetapkan nilai ke variabel. Operator ini digunakan untuk menetapkan nilai sisi kanan ekspresi ke operan sisi kiri. Simbol dasar yang digunakan adalah tanda sama dengan `=`. Beberapa contoh operator penugasan:  
    - a. Sama dengan (=).  
          Contoh : `a = 1`  
          Memberikan nilai di kanan ke dalam variabel yang berada di sebelah kiri.  
    - b. Tambah sama dengan (+=). 
          Contoh : `a += 1`  
          Memberikan nilai variabel dengan nilai variabel itu sendiri ditambah dengan nilai di sebelah kanan.  
    - c. Kurang sama dengan (-=).  
          Contoh : `a -= 1`  
          Memberikan nilai variabel dengan nilai variabel itu sendiri dikurangi dengan nilai di sebelah kanan.  
    - d. Kali sama dengan (*=).  
          Contoh : `a *= 1`  
          Memberikan nilai variabel dengan nilai variabel itu sendiri dikali dengan nilai di sebelah kanan.  
    - e. Bagi sama dengan (/=).  
          Contoh : `a /= 1`  
          Memberikan nilai variabel dengan nilai variabel itu sendiri dibagi dengan nilai di sebelah kanan.  
    - f. Sisa bagi sama dengan (%=).  
          Contoh : `a %= 1`  
          Memberikan nilai variabel dengan nilai variabel itu sendiri dibagi dengan nilai di sebelah kanan. Yang diambil nantinya adalah sisa baginya.
    - g. Pangkat sama dengan (^=).  
          Contoh : `a ^= 1`  
          Memberikan nilai variabel dengan nilai variabel itu sendiri dipangkatkan dengan nilai di sebelah kanan.  
    - h. Pembagian bulat sama dengan (//=).  
          Contoh : `a //= 1`  
          Membagi bulat operan sebelah kiri operator dengan operan sebelah kanan operator kemudian hasilnya diisikan ke operan sebelah kiri.  

6. Operator Keanggotaan (Membership).  
    Operator ini digunakan untuk mengetahui apakah suatu nilai atau variabel ada dalam koleksi data atau objek. Operator keanggotaan ini terbagi menjadi 2 yaitu:  
    - a. Operator. IN
       Dapat digunakan dengan menggunakan fungsi `in`. Opeator ini akan mengeluarkan `TRUE` jika nilai yang ditentukan ada pada koleksi data atau objek.  
    - b. Operator NOT IN.  
       Dapat digunakan dengan menggunakan fungsi `not in`. Opeator ini akan mengeluarkan `TRUE` jika nilai yang ditentukan tidak ada pada koleksi data atau objek.  

7. Operator Identitas (Identity).  
    Operator identitas digunakan untuk membandingkan objek, bukan jika sama, tetapi jika objek tersebut sebenarnya sama, dengan lokasi memori yang sama.  
    - a. Operator IS.  
       Dapat digunakan dengan fungsi `is`. Operator ini akan mengeluarkan `TRUE` jika kedua variabel adalah objek yang sama.  
    - b. Operator IS NOT.  
       Dapat digunakan dengan fungsi `is not`. Operator ini akan mengeluarkan `TRUE` jika kedua variabel bukan merupakan objek yang sama.  

8. Operator Bitwise.  
    Operator bitwise bertindak pada bit dan melakukan operasi bit demi bit. Ini digunakan untuk beroperasi pada bilangan biner. Operator Bitwise sebagai berikut:  
    - a. Bitwise AND (&).  
          Dapat dilakukan menggunakan function `&`. Operator ini mengatur setiap bit ke 1 jika kedua bit adalah 1.  
    - b. Bitwise OR (|).  
          Dapat dilakukan menggunakan fungsi `|`. Operator ini Mmengatur setiap bit ke 1 jika salah satu dari dua bit adalah 1.  
    - c. Bitwise NOT (~).  
          Dapat dilakukan menggunakan fungsi `~`. operator ini Membalikkan semua bit.
    - d. Bitwise XOR (^).  
          Adalah operasi logika dari OR Eksklusif. Kita dapat menggunakan fungsi `^`. Operator ini mengatur setiap bit ke 1 jika hanya satu dari dua bit adalah 1.
    - e. Bitwise Left Shift (<<).  
          Dapat dilakukan menggunakan function `<<`. Operator ini menggeser bit ke kiri dengan mendorong angka nol dari kanan dan biarkan bit paling kiri jatuh.
    - f. Bitwise Right Shift (>>).  
          Dapat dilakukan menggunakan fungsi `>>`. Operator ini menggeser bit ke kanan dengan mendorong salinan bit paling kiri ke dalam dari kiri, dan biarkan bit paling kanan jatuh

### 2.2.1 Program Aritmatika_2611531016
Pada praktikum ini, kode python dengan file bernama Aritmatika_2611531016 digunakan untuk melakukan operasi matematika menggunakan operator matematika yang dimana ada penjumlahan (+), pengurangan (-), pembagian (/), pembagian bulat (//), sisa bagi (%), dan pangkat (**).

````python
# Buat file dengan nama aritmatika_2611531016.py
# Buatlah program untuk operator aritmatika dalam python
# Nama variabel ditambah 4 digit nim terakhir
# Program ini menggunakan fungsi input()
# Nilai yang dimasukan akan dikonversi menjadi tipe data integer

angka1_1016 = int(input("Masukkan angka-1: "))
angka2_1016 = int(input("Masukkan angka-2: "))

# Penjumlahan
hasil_1016 = angka1_1016 + angka2_1016
print("\nOperator Penjumlahan")
print("Hasil =", hasil_1016)

# Pengurangan
hasil_1016 = angka1_1016 - angka2_1016
print("\nOperator Pengurangan")
print("Hasil =", hasil_1016)

# Perkalian
hasil_1016 = angka1_1016 * angka2_1016
print("\nOperator Perkalian")
print("Hasil =", hasil_1016)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_1016 != 0:
    hasil_1016 = angka1_1016 / angka2_1016
    print("\nOperator Pembagian")
    print("Hasil =", hasil_1016)

    hasil_1016 = angka1_1016 // angka2_1016
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_1016)

    hasil_1016 = angka1_1016 % angka2_1016
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_1016)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_1016 = angka1_1016 ** angka2_1016
print("\nOperator Pangkat")
print("Hasil =", hasil_1016)
````

### 2.2.2 Program perbandingan_2611531016

Pada praktikum ini, kode python dengan file bernama perbandingan_2611531016 digunakan untuk melakukan perbandingan suatu nilai dengan menggunakan operator perbandingan yang diantaranya ada lebih besar dari, lebih kecil dari, lebih besar atau sama dengan, lebih kecil atau sama dengan, sama dengan. tidak sama dengan, dan perbandingan berantai

code:

````python
# Buat file dengan nama perbandingan_2611531016.py
# Nama variabel ditambah 4 digit nim terakhir
# Program ini menggunakan fungsi input()
# Nilai yang dimasukan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam python

angka1_1016 = int(input("Masukkan angka-1: "))
angka2_1016 = int(input("Masukkan angka-2: "))

# Lebih besar dari
hasil_1016 = angka1_1016 > angka2_1016
print("\nOperator lebih besar dari")
print("angka1 > angka2 =", hasil_1016)

# Lebih kecil dari
hasil_1016 = angka1_1016 < angka2_1016
print("\nOperator lebih kecil dari")
print("angka1 < angka2 =", hasil_1016)

# Lebih besar dari atau sama dengan
hasil_1016 = angka1_1016 >= angka2_1016
print("\nOperator lebih besar dari atau sama dengan")
print("angka1 >= angka2 =", hasil_1016)

# Lebih kecil dari atau sama dengan
hasil_1016 = angka1_1016 <= angka2_1016
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1 <= angka2 =", hasil_1016)

# Sama dengan
hasil_1016 = angka1_1016 == angka2_1016
print("\nOperator sama dengan")
print("angka1 == angka2 =", hasil_1016)

# Tidak sama dengan
hasil_1016 = angka1_1016 != angka2_1016
print("\nOperator tidak sama dengan")
print("angka1 != angka2 =", hasil_1016)

#Tambahan: Perbandingan berantai dalam python
hasil_1016 = 0 < angka1_1016 < 100
print("\nOperator perbandingan berantai")
print("0 < angka1 < 100 =", hasil_1016)

hasil_1016 = 0 < angka2_1016 < 100
print("0 < angka2 < 100 =", hasil_1016)
````

````python
# Buat file dengan nama lainnya_2611531016.py
# Nama variabel ditambah 4 digit nim terakhir
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("===================================")
print("1. OPERATOR KEANGGOTAAN")
print("===================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_1016 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_1016 = [int(angka.strip()) for angka in input_data_1016.split(",")]

nilai_dicari_1016 = int(input("Masukkan nilai yang ingin dicari: "))

# Operator in
hasil_1016 = nilai_dicari_1016 in data_1016
print("\nOperator Keanggotaan IN")
print(nilai_dicari_1016, "in", data_1016, "=", hasil_1016)

# Operator not in
hasil_1016 = nilai_dicari_1016 not in data_1016
print("\nOperator Keanggotaan NOT IN")
print(nilai_dicari_1016, "not in", data_1016, "=", hasil_1016)


print("===================================")
print("1. OPERATOR IDENTITAS")
print("===================================")

#objek1 menggunakan list dari input pengguna
objek1_1016 = data_1016

#objek2 merujuk pada objek yang sama dengan objek1
objek2_1016 = objek1_1016

# objek3 memiliki isi yang sama, tetaoi merupakan objek baru
objek3_1016 = data_1016.copy()

print("objek1 =", objek1_1016)
print("objek2 =", objek2_1016)
print("objek3 =", objek3_1016)

# Operator is
hasil_1016 = objek1_1016 is objek2_1016
print("\nOperator Identitas IS")
print("objek1 is objek2 =", hasil_1016)

# Operator is not
hasil_1016 = objek1_1016 is not objek3_1016
print("\nOperator Identitas IS NOT")
print("objek1 is not objek3 =", hasil_1016)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =", objek1_1016 is objek3_1016)
print("objek1 == objek3 =", objek1_1016 == objek3_1016)
````

````python
# Tugas 3 Havis Lutfi
# Nim : 2611531016
# Sistem Simulasi Transaksi dan Validasi Akses Toko
# ga bikin komen takut pusing, bikin komen takut dikira ai aduhhhh
# Maaf berantakan bg puyeng mikir konsepnyah

print("\n=== SISTEM TRANSAKSI TOKO HAVISCIHUY ===\n")

nama_1016 = input("Masukkan nama Pelanggan                      : ")
member_1016 = input("Masukkan Status Pelanggan (member/nonmember) : ")
totalbelanja_1016 = int(input("Masukkan total belanja (Rp)                  : "))
jumlahbarang_1016 = int(input("Masukkan Jumlah Barang                       : "))
berat_1016= int(input("Masukkan Total Berat Barang (Kg)             : ")) 
poin_1016 = int(input("Masukkan jumlah poin (jika member) atau 0    : "))
promo_1016 = input("Masukkan Kode Promo                          : ").upper()

print("\n=== DATA TRANSAKSI ===\n")
print("Nama Pelanggan   : ", nama_1016)
print("Status Pelanggan : ", member_1016)
print("Total Belanja    : Rp", totalbelanja_1016)
print("Jumlah Barang    : ", jumlahbarang_1016)
print("Berat Total      : ", berat_1016, "Kg")
print("Kode Promo       : ", promo_1016)

kode_promo_1016 = ["HEMAT10", "HEMAT20", "GRATISONGKIR", "HAVISGANTENG" ]
# OPERATOR PERBANDINGAN
syarat_totalbelanja_1016 = totalbelanja_1016 >= 200000
syarat_jumlahbarang_1016 = jumlahbarang_1016 >= 3
statusmember_1016 = member_1016.lower() == "member"
# OPERATOR KEANGGOTAAN
promotersedia_1016 = promo_1016 in kode_promo_1016
dapatdiskon_1016 = statusmember_1016
# OPERATOR LOGIKA
dapatpromo_1016 = promotersedia_1016 and (syarat_jumlahbarang_1016 or syarat_totalbelanja_1016)

print("\n=== Hasil Validasi ===\n")
print("Belanja >= Rp200000 : ", syarat_totalbelanja_1016)
print("Jumlah Barang >= 3  : ", syarat_jumlahbarang_1016)
print("Status Member       : ", statusmember_1016)
print("Kode Promo Tersedia : ", promotersedia_1016)
print("Mendapatkan Dsikon  : ", dapatdiskon_1016)
print("Mendapatkan Promo   : ", dapatpromo_1016)

#variabel awal
diskonpromo_1016 = 0
diskonmember_1016 = 0
diskonpoin_1016 = 0
ongkir_1016 = 0
diskonongkir_1016 = 0

#harga ongkir
if berat_1016 >= 3 :       # ongkir hanya ada saat berat lebih dari atau sama dengan 3kg
    ongkir_1016 = berat_1016 * 5000 # OPERATOR ARITMATIKA

# diskon promo (takut kurang nilai kalau pake nested if)
# OPERATOR PENUGASAN DAN LOGIKA
if dapatpromo_1016 and promo_1016 == "HEMAT10":
    diskonpromo_1016 += 10000
if dapatpromo_1016 and promo_1016 == "HEMAT20":
    diskonpromo_1016 += 20000
if dapatpromo_1016 and promo_1016 == "HAVISGANTENG":
    diskonpromo_1016 += 50000
if dapatpromo_1016 and promo_1016 == "GRATISONGKIR":
    diskonongkir_1016 = ongkir_1016

# diskon poin
if poin_1016 >= 100 :
    diskonpoin_1016 = poin_1016 * 100

#diskon member
if statusmember_1016:
    diskonmember_1016 += 10000

#rumus
# OPERATOR ARITMATIKA
totaldiskon_1016 = diskonpromo_1016 + diskonmember_1016 + diskonpoin_1016
ongkirpromo_1016 = ongkir_1016 - diskonongkir_1016
totalbayar_1016 = totalbelanja_1016 - (totaldiskon_1016 + diskonongkir_1016)
ratabarang_1016 = totalbelanja_1016 / jumlahbarang_1016
dapatpoin_1016 = totalbelanja_1016 // 2000
poin_1016 += dapatpoin_1016

print("\n=== HASIL PERHITUNGAN===\n")
print("Harga Total                : Rp", totalbelanja_1016)
print("Diskon Promo               : Rp", diskonpromo_1016)
print("Diskon Member              : Rp", diskonmember_1016)
print("Diskon Poin                : Rp", diskonpoin_1016)
print("Total Diskon               : Rp", totaldiskon_1016)
print("Harga Ongkir               : Rp", ongkir_1016)
print("Harga ongkir setelah promo : Rp", ongkirpromo_1016)
print("-------------------------------------------------")
print("Total Pembayaran           : Rp", totalbayar_1016)
print("\nRata-rata Harga Barang   : Rp", ratabarang_1016)
print("Perolehan Poin             : ", dapatpoin_1016)
print("Total Poin                 : ", poin_1016)

# OPERATOR IDENTITAS
objek1_status_1016 = member_1016.lower()
objek2_status_1016 = objek1_status_1016
objek3_status_1016 = str(member_1016.lower())

is_identik_1016 = objek1_status_1016 is objek2_status_1016
is_not_identik_1016 = objek1_status_1016 is not objek3_status_1016

print("\n=== OPERATOR IDENTITAS ===")
print("objek1 is objek2     : ", is_identik_1016)
print("objek1 is not objek3 : ", is_not_identik_1016)

# OPERATOR BITWISE
if statusmember_1016:
    bitmember_1016 = 1
else:
    bitmember_1016 = 0

if syarat_totalbelanja_1016:
    bitbelanja_1016 = 2
else:
    bitbelanja_1016 = 0

if syarat_jumlahbarang_1016:
    bitbarang_1016 = 4
else:
    bitbarang_1016 = 0

if promotersedia_1016:
    bitpromo_1016 = 8
else:
    bitpromo_1016 = 0

# Operator Bitwise OR (|)
hak_akses_1016 = bitmember_1016 | bitbelanja_1016 | bitbarang_1016 | bitpromo_1016

# Operator Bitwise AND (&)
cek_member_1016 = hak_akses_1016 & 1
cek_promo_1016 = hak_akses_1016 & 8

# Operator Bitwise XOR (^)
kode_referensi_1016 = 11
beda_status_1016 = hak_akses_1016 ^ kode_referensi_1016

# Operator Bitwise Shift Left (<<)
shift_left_1016 = hak_akses_1016 << 1

print("\n=== HAK AKSES PELANGGAN ===")
print("Kode Hak Akses       : ", format(hak_akses_1016, '04b'))
print("Member Access        : ", bool(cek_member_1016))
print("Promo Access         : ", bool(cek_promo_1016))
print("Free Shipping Access : ", promo_1016 == "GRATISONGKIR")

print("\n=== OPERASI BITWISE ===")
print("=== Kode Status Transaksi ===")
print("0001 | 0010 | 0100 | 1000")
print("Kode Biner   : ", format(hak_akses_1016, '04b'))
print("Kode Desimal : ", hak_akses_1016)

print("\n=== Pemeriksaan Status ===")
print("Cek Member")
print(format(hak_akses_1016, '04b'), "& 0001")
print("Hasil Biner   : ", format(cek_member_1016, '04b'))
print("Hasil Desimal : ", cek_member_1016)

print("\nCek Promo")
print(format(hak_akses_1016, '04b'), "& 1000")
print("Hasil Biner   : ", format(cek_promo_1016, '04b'))
print("Hasil Desimal : ", cek_promo_1016)

print("\n=== Perbandingan Status ===")
print("Kode Transaksi : ", format(hak_akses_1016, '04b'))
print("Kode Referensi : ", format(kode_referensi_1016, '04b'))
print(format(hak_akses_1016, '04b'), "^", format(kode_referensi_1016, '04b'))
print("Hasil Biner   : ", format(beda_status_1016, '04b'))
print("Hasil Desimal : ", beda_status_1016)

print("\n=== Shift ===")
print(format(hak_akses_1016, '04b'), "<< 1")
print("Hasil Biner   : ", format(shift_left_1016, '05b'))
print("Hasil Desimal : ", shift_left_1016)

print("\n=== SELESAI ===")
````

Operator bitwise OR (|) menggabungkan seluruh nilai bit flag menjadi satu kode tunggal hak_akses_1016 yang menyimpan seluruh status transaksi pelanggan sekaligus. Operator bitwise AND (&) digunakan untuk memeriksa apakah bit tertentu aktif pada kode hak akses, seperti memeriksa status member dengan nilai 1 dan status promo dengan nilai 8. Operator bitwise XOR (^) membandingkan nilai bit pada kode hak akses dengan kode_referensi_1016 sebesar 11 untuk menghasilkan bit 1 jika terdapat perbedaan kondisi status di antara keduanya. Operator bitwise Shift Left (<< 1) menggeser seluruh bit pada kode hak akses sebanyak satu posisi ke kiri, yang secara matematis mengalikan nilai desimal dari hak akses tersebut dengan dua.
