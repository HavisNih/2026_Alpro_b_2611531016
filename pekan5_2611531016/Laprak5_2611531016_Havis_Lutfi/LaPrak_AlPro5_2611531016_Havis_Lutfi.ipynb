## 2.1 Dasar Teori
Pada praktikum ini mahasiswa melakukan pengenalan dan mengimplementasikan tentang perulangan for dan nested for.

1. Perulangan.
    Perulanngan atau loop digunakan untuk mengeksekusi blok kode berulang kali hingga suatu kondisi terpenuhi atau semua item dalam suatu urutan diproses. Jenis utamanya adalah for loop (mengulangi urutan) dan While loop (mengeksekusi kode berdasarkan suatu kondisi).
2. Perulangan for.
    Perulangan for digunakan untuk mengulangi suatu urutan (yaitu daftar, tupel, kamus, himpunan, atau string). Dengan perulangan for kita dapat mengeksekusi sekumpulan pernyataan, satu kali untuk setiap item dalam daftar, tupel, kumpulan, dll. Kita bisa menggunakanny dengan menggunakan pernyataan `for ... in range (....)`
3. Nested for loop.
    Bahasa pemrograman Python memungkinkan penggunaan satu lingkaran di dalam loop lain. Bagian berikut menunjukkan beberapa contoh untuk menggambarkan konsep tersebut. Kita membuat nested for dengan menggunakan perulangan for dalam perulangan for. 
4. range(start, stop)
    stop adalah batas akhir yang tidak diikutsertakan. Jadi perulangannya sebenarnya berhenti ketika mencapai nilai stop, namun nilai terakhir yang dijalankan/dicetak adalah stop - 1.
    - 1 Nilai `range(stop)`: 
        Jika hanya diisi 1 nilai, nilai tersebut dianggap sebagai stop. Perulangan otomatis dimulai dari 0 dan angka terakhir yang diproses adalah stop - 1
    - 2 Nilai `range(start, stop)`: 
        Jika diisi 2 nilai, nilai pertama adalah start (titik awal) dan nilai kedua adalah stop (batas akhir). Perulangan dimulai dari start hingga angka terakhir stop - 1.

### 2.2.5 nested_for1_2611531016

````python
# Buat file dengan nama nested_for1_2611531016.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir
# Program ini menggunakan fungsi input()

batas_1016 = int(input("Masukkan nilai batas: "))
for line_1016 in range(1, batas_1016 + 1):
    for j_1016 in range(1, (-1 * line_1016 + batas_1016) + 1):
        print(".", end="")
    print(line_1016)
````

### 2.2.8 Program nested_for4_2611531016

````python
# Buat file dengan nama nested_for4_2611531016.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir
# Program ini menggunakan fungsi input()

tinggi_1016 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1016 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_1016 = tinggi_1016
    c_1016 = a_1016
    lebar_1016 = (2 * tinggi_1016) - 2

    for i_1016 in range(1, tinggi_1016 + 1):
        b_1016 = c_1016 + 1

        for j_1016 in range(1, lebar_1016 + 1):

            # Baris atas dan bawah
            if i_1016 == 1 or i_1016 == tinggi_1016:
                if j_1016 == 1 or j_1016 == lebar_1016:
                    print("#", end="")
                else:
                    print("=", end="")

                    # Baris isi
            else:
                if j_1016 == 1 or j_1016 == lebar_1016:
                    print("|", end="")
                else:
                    if j_1016 == c_1016:
                        print("<", end="")
                    elif j_1016 == b_1016:
                        print(">", end="")
                    elif j_1016 == (lebar_1016 - c_1016):
                        print("<", end="")
                    elif j_1016 == (lebar_1016 - c_1016 + 1):
                        print(">", end="")
                    elif j_1016 > b_1016 and j_1016 < (lebar_1016 - c_1016):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_1016 -= 2

        if a_1016 <= 0:
            c_1016 = (-a_1016) + 2
        else:
            c_1016 = a_1016
````

4. Loop utama
